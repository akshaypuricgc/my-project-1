```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data = {
    "Physical_Activity": [30, 60, 20, 80, 50],
    "Stress_Level": [8, 3, 7, 2, 5],
    "Sleep_Hours": [5, 8, 6, 9, 7],
    "Heart_Rate": [85, 70, 80, 65, 75],
    "Daily_Steps": [3000, 10000, 4000, 12000, 7000]
}
df = pd.DataFrame(data)
print("Dataset:\n")
print(df)
```

```python
print("\nDataset Info:\n")
print(df.info())
```

```python
print("\nStatistical Summary:\n")
print(df.describe())
```

```python
print("\nCorrelation Matrix:\n")
print(df.corr())
```

```python
import seaborn as sns

plt.figure()
sns.heatmap(df.corr(), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure()
plt.scatter(df["Physical_Activity"], df["Sleep_Hours"])
plt.xlabel("Physical Activity")
plt.ylabel("Sleep Hours")
plt.title("Physical Activity vs Sleep")
plt.show()
```

```python
plt.figure()
plt.scatter(df["Stress_Level"], df["Sleep_Hours"])
plt.xlabel("Stress Level")
plt.ylabel("Sleep Hours")
plt.title("Stress vs Sleep")
plt.show()
```

```python
plt.figure()
plt.scatter(df["Heart_Rate"], df["Sleep_Hours"])
plt.xlabel("Heart Rate")
plt.ylabel("Sleep Hours")
plt.title("Heart Rate vs Sleep")
plt.show()
```

```python
plt.figure()
plt.scatter(df["Daily_Steps"], df["Sleep_Hours"])
plt.xlabel("Daily Steps")
plt.ylabel("Sleep Hours")
plt.title("Steps vs Sleep")
plt.show()
```

```python
print("\nConclusion:")
print("Sleep is proportional to healthy lifestyle.")
print("More physical activity and steps improve sleep.")
print("Higher stress reduces sleep quality.")
print("Lower heart rate indicates better fitness and better sleep.")
```
